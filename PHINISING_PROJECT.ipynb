```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import seaborn as sns

raw_mail_data = pd.read_csv('/content/mail_data.csv')
```

```python
print(raw_mail_data)
```

```python
data=raw_mail_data.where((pd.notnull(raw_mail_data)),'')
data.head()
```

```python
data.shape
```

labelling the data


```python
data.loc[data['Category']=='spam','category',]=0
data.loc[data['Category']=='ham','category',]=1
```

making x and y axis

```python
x=data['Message']
y=data['category']
```

```python
print(x,y)#check
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=3)
```

feature extraction

```python
# transform the text data to feature vectors that can be used as input to the Logistic regression

feature_extraction = TfidfVectorizer(min_df = 1, stop_words='english', lowercase=True)

xtrain_features = feature_extraction.fit_transform(xtrain)
xtest_features = feature_extraction.transform(xtest)

# convert Y_train and Y_test values as integers

ytrain = ytrain.astype('int')
ytest = ytest.astype('int')
```

```python
print(xtest_features)
```

done with data tuning time for making the model

```python
model=LogisticRegression()
```

```python
model.fit(xtrain_features,ytrain)
```

MODEL PERFORMANCE VIEWING

```python
prediction_ontrainingdatas=model.predict(xtrain_features)
accuracy_ontrainingdatas=accuracy_score(ytrain,prediction_ontrainingdatas)
```

```python
print("accuracy for trained dataset",accuracy_ontrainingdatas)
```

MODEL PERFORMANCE ON TEST DATA

```python
prediction_ontrainingdatas=model.predict(xtest_features)
accuracy_ontrainingdatas=accuracy_score(ytest,prediction_ontrainingdatas)
print("accuracy for testing dataset",accuracy_ontrainingdatas)
```

the copy paste part

```python
enter_the_mail=["As per your request 'Melle Melle (Oru Minnaminunginte Nurungu Vettam)' has been set as your callertune for all Callers. Press *9 to copy your friends Callertune"]
transformit=feature_extraction.transform(enter_the_mail)
prediction=model.predict(transformit)
print(prediction)
if(prediction[0]==1):
  print("ham mail")
else:
  print("spam mail")
```

saving the trained model as a file for building UI

```python
import joblib
joblib.dump(model, 'spam_classifier_model.pkl')
joblib.dump(feature_extraction, 'tfidf_vectorizer.pkl')

```
